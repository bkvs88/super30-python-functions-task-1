Create a function that accepts a string and determines whether it is a palindrome.

```python
def accept_data(name,age,course):  # defining the function with 3 inputs
    return name,age,course # return the inputs
```

```python
accept_data("sravan",36,"Python") # calling the function with actual sequence
```

```python
accept_data(age=36,course="Python",name="sravan") # calling the function with positional arguments
```
